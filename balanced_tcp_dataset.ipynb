```python
import pandas as pd
```

```python
normal = pd.read_csv("normal_tcp_big.csv")
syn = pd.read_csv("syn_flood.csv")
ack = pd.read_csv("ack_flood.csv")
fin = pd.read_csv("fin_flood.csv")
```

```python
normal['label'] = 0
syn['label'] = 1
ack['label'] = 2
fin['label'] = 3
```

```python
min_size = 5000

normal_sampled = normal.sample(n=min_size, random_state=42)
syn_sampled = syn.sample(n=min_size, random_state=42)
ack_sampled = ack.sample(n=min_size, random_state=42)
fin_sampled = fin.sample(n=min_size, random_state=42)
```

```python
final_df = pd.concat([
    normal_sampled,
    syn_sampled,
    ack_sampled,
    fin_sampled
])
```

```python
final_df = final_df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
final_df.to_csv("final_multiclass_dataset.csv", index=False)
```
